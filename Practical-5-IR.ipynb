Step 1: Import NetworkX

```python
import networkx as nx
```

Step 2: Create the Graph

```python
G = nx.barabasi_albert_graph(60, 41)

print("Number of Nodes:", G.number_of_nodes())
print("Number of Edges:", G.number_of_edges())
```

Step 3: Calculate PageRank

```python
pr = nx.pagerank(G, alpha=0.4)

print("PageRank Values:")
print(pr)
```

Step 4: Display PageRank in Descending Order

```python
sorted_pr = sorted(pr.items(), key=lambda x: x[1], reverse=True)

print("PageRank in Descending Order:\n")

for node, score in sorted_pr:
    print(f"Node {node}: {score:.6f}")
```

Step 5: Find the Highest PageRank Node

```python
highest_node = max(pr, key=pr.get)

print("\nHighest PageRank Node:", highest_node)
print("Highest PageRank Score:", pr[highest_node])
```

Step 6: Visualize the Graph

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
nx.draw(G, with_labels=True, node_size=500)
plt.title("Barabasi-Albert Graph")
plt.show()
```

```python
# --- 1. CODE EXPLANATION ---
#
# Libraries:
# - networkx (nx): Used to create, manipulate, and analyze complex graph networks.
# - pylab (plt): Used for plotting and visualizing the generated graph.
#
# Graph Generation:
# - nx.barabasi_albert_graph(60, 41): Generates a scale-free network with 60 nodes, 
#   where each new node attaches to 41 existing nodes based on preferential attachment 
#   (nodes with more connections are more likely to receive new links).
#
# PageRank Calculation:
# - nx.pagerank(G, alpha=0.4): Computes the importance/rank of each node in graph G.
# - alpha=0.4 (Damping Factor): Represents the probability that a random walker continues 
#   clicking links rather than jumping to a random node (1 - alpha = 0.6 jump probability).
#   Note: Standard web PageRank typically uses alpha = 0.85.
#
# Output & Visualization:
# - print("Page Rank Value is:", pr): Outputs a dictionary mapping each node (0 to 59) 
#   to its calculated PageRank score (all scores sum up to 1.0).
# - nx.draw(G): Renders a visual representation of the network nodes and edges.


# --- 2. VIVA / ORAL EXAM Q&A ---
#
# Q1: What is the PageRank algorithm?
# A1: PageRank is a link-analysis algorithm used to measure the relative importance of 
#     nodes within a graph. A node gets a higher rank if it is linked to by many other 
#     nodes, especially if those linking nodes themselves have high ranks.
#
# Q2: What does the damping factor (alpha = 0.4) represent?
# A2: It represents the probability of continuing to follow edges in the graph. Setting 
#     alpha to 0.4 means there is a 40% chance of following a link and a 60% (1 - 0.4) 
#     chance of randomly jumping to any other node in the network.
#
# Q3: Why did you use the Barabási-Albert graph model here?
# A3: The Barabási-Albert model generates scale-free networks using "preferential attachment" 
#     (rich-get-richer mechanism). This closely mimics real-world networks like the World 
#     Wide Web, where a few hub nodes have many links and most nodes have very few.
#
# Q4: What is the sum of all PageRank values in the output dictionary?
# A4: The sum of all PageRank scores across all nodes in the graph always equals 1.0, 
#     as PageRank represents a stationary probability distribution.
#
# Q5: What happens if alpha is set to 0 or 1?
# A5: If alpha = 0, every node gets an equal rank (1/N) because navigation is purely random. 
#     If alpha = 1, the random walker never jumps, which can cause rank sinks (getting 
#     trapped in dead-end nodes or closed loops).
```
