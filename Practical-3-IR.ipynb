```python
import numpy as np
import pandas as pd

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator

# Load dataset
heartDisease = pd.read_csv("heart.csv")

# Clean column names
heartDisease.columns = heartDisease.columns.str.strip()

# Handle missing values
heartDisease = heartDisease.replace("?", np.nan)
heartDisease = heartDisease.dropna()

# Rename last column as heartdisease
if "heartdisease" not in heartDisease.columns:
    heartDisease = heartDisease.rename(
        columns={heartDisease.columns[-1]: "heartdisease"}
    )

print("Dataset columns:")
print(heartDisease.columns.tolist())

# Create Bayesian Network
model = DiscreteBayesianNetwork([
    ("age", "trestbps"),
    ("age", "fbs"),
    ("sex", "trestbps"),
    ("exang", "trestbps"),
    ("trestbps", "heartdisease"),
    ("fbs", "heartdisease"),
    ("heartdisease", "restecg"),
    ("heartdisease", "thalach"),
    ("heartdisease", "chol")
])

# Learn CPDs
estimator = MaximumLikelihoodEstimator(model, heartDisease)
cpds = estimator.get_parameters()

for cpd in cpds:
    model.add_cpds(cpd)

print("\nBayesian Network created successfully.")
print("Model check:", model.check_model())

print("\nLearned CPDs:")
for cpd in model.get_cpds():
    print(cpd)
```

```python
from pgmpy.inference import VariableElimination

# Create inference object
heartDisease_infer = VariableElimination(model)

# Diagnosis using age = 63
print("\n--- Diagnosis using Age = 63 ---")

q1 = heartDisease_infer.query(
    variables=["heartdisease"],
    evidence={"age": 63}
)

print(q1)


# Diagnosis using cholesterol = 233
print("\n--- Diagnosis using Cholesterol = 233 ---")

q2 = heartDisease_infer.query(
    variables=["heartdisease"],
    evidence={"chol": 233}
)

print(q2)
```

```python
# ==============================================================================
# EXTERNAL VIVA QUESTIONS & ANSWERS (COMPLETE PRACTICAL)
# ==============================================================================

# Q1: What is a Bayesian Network?
# A1: A Bayesian Network (or Belief Network) is a Probabilistic Graphical Model 
#     represented as a Directed Acyclic Graph (DAG), where nodes represent random 
#     variables (such as age, cholesterol, heartdisease) and directed edges represent 
#     conditional dependencies between them, quantified by Conditional Probability 
#     Distributions (CPDs)[cite: 3].

# Q2: Why must a Bayesian Network be a Directed Acyclic Graph (DAG)?
# A2: It must be acyclic (no closed loops) so that joint probabilities can be 
#     cleanly factorized using the Chain Rule of Probability without circular 
#     dependencies, ensuring exact inference algorithms converge properly.

# Q3: What is a CPD (Conditional Probability Distribution) in this code?
# A3: A CPD (or CPT - Conditional Probability Table for discrete variables) 
#     specifies the probability of every possible state of a node given every 
#     possible combination of states of its parent nodes[cite: 3]. For root nodes 
#     without parents (like `age` or `sex`), it stores prior marginal probabilities[cite: 3].

# Q4: How are the CPDs learned from data in this practical?
# A4: Using Maximum Likelihood Estimation (`MaximumLikelihoodEstimator`), which 
#     estimates conditional probabilities directly from the relative frequencies 
#     (counts) of variable combinations occurring in `heart.csv`[cite: 3].

# Q5: What does `model.check_model()` validate?
# A5: It checks two things: (1) every node in the DAG has an associated CPD 
#     matching its parent structure, and (2) the probabilities in each column of 
#     every CPD sum to 1.0[cite: 3].

# Q6: What is Variable Elimination, and why is it used in Cell 13?
# A6: `VariableElimination` is an exact inference algorithm used to compute the 
#     posterior probability distribution of query variables (e.g., `heartdisease`) 
#     given observed evidence (e.g., `age=63` or `chol=233`) by summing out 
#     (marginalizing) all non-query, non-evidence variables efficiently[cite: 3].

# Q7: Notice that `chol` is a child of `heartdisease` (`heartdisease -> chol`), yet 
#     we queried `heartdisease` given `chol = 233`. How is that possible?
# A7: Bayesian Networks support both causal (top-down) and diagnostic (bottom-up) 
#     reasoning using Bayes' Theorem. Even though the arrow points from 
#     `heartdisease` to `chol`, observing `chol = 233` updates our belief about 
#     its parent `heartdisease` via Bayesian inversion[cite: 3].

# Q8: Why did we use `DiscreteBayesianNetwork` instead of a continuous model, and 
#     what is a limitation of using raw continuous values like `age` or `chol` here?
# A8: `DiscreteBayesianNetwork` treats every unique numeric value as a separate 
#     discrete category[cite: 3]. Because `age`, `trestbps`, `thalach`, and `chol` have 
#     many distinct integer values, their CPDs become very large and sparse[cite: 3] 
#     (some combinations have 0 samples). Discretizing/binning continuous columns 
#     into ranges (e.g., Low/Normal/High) improves generalization.

# Q9: How does a Bayesian Network differ from a Naive Bayes classifier?
# A9: Naive Bayes assumes all features are conditionally independent of each other 
#     given the class label (a fixed star-shaped graph). A Bayesian Network allows 
#     arbitrary, realistic causal relationships between features themselves (such as 
#     `age -> trestbps` and `sex -> trestbps`)[cite: 3].

# Q10: Why did we replace `'?'` with `np.nan` and call `dropna()` during preprocessing?
# A10: In the standard UCI Heart Disease dataset, unknown values are recorded as 
#      `'?'` strings. Converting them to `NaN` allows `pd.to_numeric()` to parse 
#      columns as numbers and `dropna()` to remove incomplete records before 
#      computing frequency counts for CPDs[cite: 3].
# ==============================================================================
```
